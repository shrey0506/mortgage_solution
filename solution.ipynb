```python
# 1.There are missing values observed in the dataset.


# 2. One column is datetime, and 3 are obj/categorical columns and rest are float/numeric columns.


# 3. Product name has missing values which suggests, user might have clicked on the product but it might not have been captured.
#    Hence dropped the values of product, as it might be mapped to the wrong order.


# 4. Since we don't know here what was the core reason to put product_postion at specifiec place it will be risky to impute the values(
#    example rank 1 at 100 clicks might not make any sense however rank 1 within<5 clicks it makes sense, which is also vague to interpret).


# 5. page level session - We don't want to capture the unknown click, as it is totally on the user. As mentioned even if it is the best product
#    It is not guaranteed, this click will be good, or user is going to spent good time in that session.
```

```python
df_copy = df.copy()
```

```python
df_clean = df_copy.dropna(subset=['Product_Name', 'product_position', 'page_level_sessions'])
```

```python
production_position_at_zero = df_clean[df_clean['product_position'] == 0]
print(production_position_at_zero.shape)
production_position_at_zero.head()
```

```python
# clearly we can observe the data has those values whose product_position is 0 which doesnot make sense.
# Hence we will capture those data points whose product_position>0
```

```python
df_clean = df_clean[df_clean['product_position'] > 0]
```

```python
# tier_revenue_per_conversion is missing meaning Realized revenue is missing, meaning user dropped out hence imputing it with 0
fill_na_with_zero_cols = ['tier_revenue_per_conversion','tier_revenue_per_clickout']
df_clean[fill_na_with_zero_cols] = df_clean[fill_na_with_zero_cols].fillna(0)
```
